```python
marks = []
for i in range(1, 5):
    while True:
        try:
            mark = int(input(f"Enter mark for subject {i}: "))
            if 0 <= mark <= 100:
                marks.append(mark)
                break
            else:
                print("Invalid mark: Mark should be between 0 and 100. Please try again.")
        except ValueError:
            print("Invalid input: Please enter a numeric value for the mark. Please try again.")

if len(marks) == 4:
    total_marks = sum(marks)
    percentage = (total_marks / 400) * 100
    print(f"\nPercentage: {percentage:.2f}%")

    if percentage >= 90:
        print("Grade A")
    elif percentage >= 80:
        print("Grade B")
    elif percentage >= 70:
        print("Grade C")
    elif percentage >= 60:
        print("Grade D")
    else:
        print("Grade F")
else:
    print("An error occurred in collecting marks. Please re-run the cell.")
```
