```python
num = int(input("Enter a number: "))
original = num
total = 0
while num > 0:
  digit = num % 10
  total += digit
  num //= 10
print(f"sum of digits of {original} = {total}")
```

```python
num = int(input("Enter a number: "))
count = 0
while count <= 10:
  table = count * num
  print(f"{count} x {num} = {table}")
  count += 1

```
