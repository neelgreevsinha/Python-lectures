```python
# Q2
word = input("Enter a word:")
vowels = "aeiouAEIOU"
v = []

for letter in word:
  if letter in vowels:
    v.append(letter)

v.reverse()

answer = ""
j = 0

for letter in word:
  if letter in vowels:
    answer = answer + v[j]
    j = j + 1
else:
   answer = answer + letter
print(answer)
```
