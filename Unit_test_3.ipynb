```python
# Question 2
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a



def lcm(a, b):
    return abs(a * b) // gcd(a, b)



def gcd_rec(a, b):
    if b == 0:
        return a
    return gcd_rec(b, a % b)



pairs = [(48, 18), (100, 75), (17, 5)]


print(f"{'a':<10}{'b':<10}{'GCD':<10}{'Recursive GCD':<18}{'LCM':<10}")
print("-" * 58)

for a, b in pairs:
    print(f"{a:<10}{b:<10}{gcd(a, b):<10}{gcd_rec(a, b):<18}{lcm(a, b):<10}")
```
